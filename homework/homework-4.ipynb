# Homework 4: Parsing Rotten Tomatoes reviews

In this homework, you have to create a script that will parse data from [Rotten Tomatoes](https://rottentomatoes.com), a movie reviews website. 
- The work you have to do is similar to what we covered in Lectures 4-6, albeit for a different website. 
- Please read and follow the instructions below very carefully.

## Step 1
---

Your script should begin by importing libraries, and then defining two variables
1. **movie** a string variable indicating the movie for which reviews will be parsed
2. **pageNum** a positive whole number indicating how many batches of reviews to parse

For example, to parse the first 3 batches of the [Gangs of New York](https://www.rottentomatoes.com/m/gangs_of_new_york/reviews) reviews, I would have to open your script and set **movie='gangs_of_new_york'** and **pageNum=3**. 

Your code should then go to the movie's [All Critics](https://www.rottentomatoes.com/m/gangs_of_new_york/reviews/all-critics) reviews page, and parse the first 3 batches of reviews.
- Pagination on RT happens by clicking on the "Load More" button. `pageNum=1` means the reviews that load first. `pageNum=3` means those reviews, plus up to two successful clicks on Load More.
- Do not save the same review twice. If there are fewer batches than requested, save the reviews that are available.
- If you get stuck, email the TA.

## Step 2
---

For each review contained in each of the batches you requested, parse the following information
1. **The critic** This should be None if the review doesn't have a critic's name.
2. **The rating.** The rating should be 'rotten', 'fresh', or None if the review doesn't have a rating.
3. **The source** This should be None if the review doesn't have a source.
4. **The text.**  This should be None if the review doesn't have text.
5. **The date.**  This should be None if the review doesn't have a date.

## Step 3

After parsing the data, save them in a file that is called **firstname_lastname_movie.json**,

The JSON file should contain a list of dicts, one for each review, with exactly these five keys. Here is a valid JSON example with made-up values:

```json
[
    {
        "critic_name": "Example Critic",
        "rating": "fresh",
        "source": "Example Publication",
        "text": "An enjoyable film.",
        "date": "Oct 1, 2026"
    },
    {
        "critic_name": null,
        "rating": null,
        "source": "Another Publication",
        "text": "An interesting idea.",
        "date": null
    }
]
```

**Note:** 
- [This notebook](https://github.com/apostolosfilippas/wa/blob/main/scripts/06.1%20JSON.ipynb) introduces you to JSON. For this homework, we save a list of dicts, all with the same keys.
- Use `None` for missing information in your Python code. `json.dump` turns it into `null` in the JSON file. Do not use the string `"NA"`.
- The rating should be `"fresh"`, `"rotten"`, or `None`, not a numerical score such as `6/10`.
