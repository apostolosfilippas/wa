# Homework 6: Build a game

In this homework, you will build a **small, playable game**. You can use **any programming language** you want. The type of game is up to you too. It could be a puzzle, a trivia game, an arcade game, a text adventure, or something completely different.

I'll play your games in class, and then we'll vote on the best one!

You can work individually or in a group of **up to three people**. I expect more from larger groups, both in what the game does and how well it works.

## Your game

Give the player a clear goal and explain how to play. Keep the scope small enough that you can finish it and test it properly. A simple idea can make a great game!

Think about what would make someone want to play again. Try it yourself, and have someone else play it too. See where they get confused or stuck, and use that to improve it.

## Working with an AI agent

Start by clearly telling the agent what you want to build. Explain the game idea, what the player should be able to do, and how you imagine it looking.

Go back and forth to work out the details, including which language and tools to use. Ask the agent to explain its suggestions in simple terms so you understand what you're choosing.

**Get to a working graphical interface quickly!** Ask for a simple first version that you can open and play. Try it, see what works, and tell the agent what you'd like to change. Be specific about what happened and what you expected. Screenshots can help too.

Keep going back and forth. Make a change, play again, and see whether it improved the game.

## Deliverables

Put your game in a **GitHub repository** and invite me (**@apostolosfilippas**) as a collaborator. Submit the repository link. Include the following in your repository.

1. **Your game code**, including any images, sounds, or other files needed to run it.
2. **A short README** with the game's name, your name(s), the rules and controls, and instructions for starting it. Include any setup steps or packages we need. If the game runs on a website, include a link we can open and play.

Before submitting, follow your own instructions from the beginning and check that the game runs. We should be able to get it running and start playing during class.
