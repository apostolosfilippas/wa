# ⚪⚫ 1. Sets


Another useful data type is the **set**.


Sets are an **unordered** collection of **unique** items. Let's see how they work through examples.

```python
# Use set() to create an empty set. {} creates an empty dictionary.
my_set = set()
my_set
```

Our set is empty, so let's start adding items to it

```python
my_set.add(0)
my_set
```

```python
my_set.add(1)
my_set.add(2)
my_set
```

```python
# now let's add a duplicate
my_set.add(2)

# notice that the duplicate is not added - sets contain unique values
my_set
```

```python
# `update` lets you add multiple elements at once
my_set.update([11,12,13])
my_set
```

```python
# `remove` lets you remove an element
my_set.remove(11)
my_set  
```

```python
#  but will throw an error if the element is not in the set
my_set.remove(11)
```

```python
# `discard` lets you remove an element without throwing an error if it's not in the set
my_set.discard(9)
```

```python
# A set can contain different types of elements, such as numbers and strings.
my_set.add("AF")
my_set
```

### Casting

We can also change lists to sets and back, a process called casting. Let's see how this works.

One thing to keep in mind: `set(my_list)` removes duplicates, but it does not preserve the list's order.

Use `list(my_set)` to convert a set back to a list. This will not restore the original order or any duplicates.

```python
my_list = [1,2,3,4,0,2,3,4,5,6,1,2,3]
my_list
```

```python
# Convert the list to a set to keep only unique values.
set(my_list)
```

### Other set operations

Going back to our math education, let's look at three useful operations:

- **Union:** elements in either set (including elements in both).
- **Intersection:** elements in both sets.
- **Difference:** elements in the first set but not the second.

Let's see some examples. Notice that these methods return a new set; they do not change A or B.

```python
A = {0,1,3,5,9,10}
B = {0,2,4,6,8,10}
```

```python
# `union` returns all elements in either set
A.union(B)
```

```python
# therefore it does not matter which set we call the method on
B.union(A)
```

```python
# `intersection` returns only elements in both sets
A.intersection(B)
```

```python
# again, it does not matter which set we call the method on
B.intersection(A)
```

```python
# `difference` returns a set containing all the elements of the first set that are not in the second set.
A.difference(B)
```

```python
# Difference depends on which set comes first.
B.difference(A)
```

### Sets vs. lists

1. **Uniqueness:** sets contain unique elements; lists can contain duplicates.
2. **Order and indexing:** lists preserve their order and support indexing, such as `my_list[0]`. Sets have no first or second element, so `my_set[0]` does not work. Do not rely on the order in which a set is displayed.
3. **Allowed elements:** both lists and sets can be changed. The items inside a set must be **hashable**: Python needs a stable hash value for each item so it can look it up. For the types we've seen, numbers and strings work, but lists, dictionaries, and sets do not. A tuple works only if all of its elements are hashable: `(1, 2)` works, but `(1, [2, 3])` does not.

### Example: website visitors

Imagine you're a web analyst for an e-commerce company. You've just had a big sale on your website, and you've collected visitor IDs for two days: the day of the sale and the day after. You want to know:

- How many unique visitors came each day?
- How many visitors came on both days, and who were they?
- How many visitors came only on the sale day, and who were they?

Sets are perfectly suited for this. Let's see how!

```python
# Day of the sale
sale_day_visitors = ["user1", "user2", "user3", "user4", "user2", "user5"]

# Day after the sale
post_sale_day_visitors = ["user4", "user6", "user7", "user8", "user1"]

# Converting lists to sets to get unique visitors
sale_day_set = set(sale_day_visitors)
post_sale_day_set = set(post_sale_day_visitors)
```

### Checking membership

Was a particular user in our set? We can check with `in`. And, as you'd expect, `not in` checks the opposite!

```python
print("user1" in sale_day_set)      # True: user1 visited on the sale day.
print("user6" not in sale_day_set)  # True: user6 did not visit on the sale day.
```

```python
# 1. Number of unique visitors each day
print(f"Unique visitors on sale day: {len(sale_day_set)}")
print(f"Unique visitors on the day after the sale: {len(post_sale_day_set)}")
```

```python
# 2. Count and identify visitors who came on both days.
both_days = sale_day_set.intersection(post_sale_day_set)
print(f"Number of visitors on both days: {len(both_days)}")
# sorted() returns a list so the IDs are displayed in a consistent order.
print(f"Visitor IDs: {sorted(both_days)}")
```

```python
# 3. Count and identify visitors who came only on the sale day.
only_sale_day = sale_day_set.difference(post_sale_day_set)
print(f"Number of visitors only on the sale day: {len(only_sale_day)}")
print(f"Visitor IDs: {sorted(only_sale_day)}")
```

# 🔗 ✂️ 2. Assignment and copying

When we write `y = x`, what actually happens? Do we get a new copy of x?

No! A variable is a name that refers to an object. The `=` operator makes a name refer to an object; it does **not** copy it.

So after `y = x`, both names refer to the same object. If we change that object through y, we'll see the change through x too.

Let's see an example.

```python
x = [1, 2, 3]
y = x  # Both names refer to the same list.

y.append(4)  # Change the shared list.
print(y)
```

```python
# The change is visible through x too.
x
```

### Changing an object vs. reassigning a name

Notice the difference here: `y.append(4)` changes the list itself. But `y = [4]` makes y refer to a new list. It doesn't change the list that x refers to!

```python
y = [4]  # Reassign y; do not change the list that x refers to.
print(f"x: {x}")  # [1, 2, 3, 4]
print(f"y: {y}")  # [4]
```

```python
my_dict = {'cat': 'die Katze', 'dog': 'der Hund'}

# Give the same dictionary another name; this does not create a copy.
second_dict = my_dict
```

```python
# == compares values. Both names currently refer to the same dictionary.
print(my_dict == second_dict)
print(my_dict)
print(second_dict)
```

The `is` operator checks if two variables refer to the exact same object in memory, not just if they are equal.

```python
# is checks whether both names refer to the same object.
my_dict is second_dict
```

```python
# Change the shared dictionary; the change is visible through both names.
my_dict['mice'] = 'die Mäuse'
print(my_dict)
print(second_dict)
```

```python
del my_dict['mice']
print(my_dict)
print(second_dict)
```

### Making a copy

The variables `my_dict` and `second_dict` refer to the same dictionary!

This can be useful, but also inconvenient. What if we want to change the second dictionary without changing the first? We can make a copy with `my_dict.copy()`.

This gives us a new dictionary with the same entries. But there's a catch: the objects stored inside are still shared. This is called a **shallow copy**. We'll see why this matters in a bit.

```python
second_dict = my_dict.copy()
print(my_dict == second_dict)  # True: their contents are equal.
print(my_dict is second_dict)  # False: they are different dictionaries.
```

The same thing holds for lists and sets: they also have a `.copy()` method. We can also use `dict(my_dict)` or `list(my_list)` to make shallow copies.

Now see what happens when we add an entry to the original dictionary:

```python
my_dict['the butterfly'] = 'der Schmetterling'

# The new entry appears only in the original dictionary.
print(my_dict)
print(second_dict)
```

### Nested objects and shallow copies

Let's create a dictionary with a list inside it. Notice that both `'c'` and `'d'` refer to the same list as x:

```python
x = [5, 6, 7]

my_dict = {
    'a': 1,
    'b': 2,
    'c': x,
    'd': x
}

print(my_dict)
```

Now let's change the list that x refers to. What happens to our dictionary?

```python
x.append(8)  # Change the list shared by x, my_dict['c'], and my_dict['d'].
print(my_dict)
```

Notice that both `'c'` and `'d'` show the change, because they refer to the same list.

And here's the catch we mentioned earlier: copying the dictionary does **not** copy the list inside it. Let's see:

```python
copied_dict = my_dict.copy()
print(copied_dict is my_dict)            # False: different outer dictionaries.
print(copied_dict['c'] is my_dict['c'])  # True: the nested list is shared.

copied_dict['c'].append(9)  # Change the shared list, not just the copy.
print(my_dict['c'])        # [5, 6, 7, 8, 9]
print(copied_dict['c'])    # [5, 6, 7, 8, 9]
```

### What to remember

- `y = x` gives an object another name; it does not copy it.
- Changing a shared object is different from reassigning one of its names.
- `.copy()` creates a new outer container, but nested objects can still be shared.

**For those of you who are curious:** if we want to copy the objects inside too, we can use `copy.deepcopy()`. You can read more in [Python's copying documentation](https://docs.python.org/3/library/copy.html).
