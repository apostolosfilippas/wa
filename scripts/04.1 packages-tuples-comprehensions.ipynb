# 📦 1. Packages
A **module** is code that someone has already written for us to use. It might be a `.py` file, or it might be built into Python. A **package** groups related modules together. People often use the two words loosely, and that's okay for now.
Packages and modules let us avoid reinventing the wheel every time we want to do something in Python. They can save us a lot of time!

## Importing modules that come with Python
Python comes with a **standard library**: useful modules that you do not have to install separately. `math` is one of them.

To use it, write `import math`. Python loads it the first time you import it in a running program or notebook. Then, the dot in `math.sqrt` tells Python: look inside `math` for the function called `sqrt`.

Let's see a common example.

```python
import math

# Calculate the square root
sqrt_25 = math.sqrt(25)
print(f"Square root of 25 is: {sqrt_25}")

# Calculate the factorial
factorial_5 = math.factorial(5)
print(f"Factorial of 5 is: {factorial_5}")

# Calculate the cosine of an angle
cos_45 = math.cos(math.radians(45))  # Convert degrees to radians first
print(f"Cosine of 45 degrees is: {cos_45}")
```

Here `sqrt(25)` gives us the square root of 25. `factorial(5)` means 5 × 4 × 3 × 2 × 1. And `cos` expects an angle in **radians**, so we use `math.radians(45)` to convert 45 degrees first.

The name before the dot matters: if we had not imported `math`, Python would not know what `math.sqrt` means.

Pretty cool right? Now all of the code that so many people have written in the past is available for you to use!

That is in large part the power of programming and of open source!

# 🧬 2. Tuples



Tuples group several objects together. Think of them as lists that you do not intend to change.
- Once you create a tuple, you cannot replace, add, or remove its positions. Python calls this **immutable**.
- This can be useful when a group of values should stay together, like a product name and its price.

A tuple is **ordered**: position 0 is the first item, just like in a list. It can also hold different types of values at once.

We usually write tuples with parentheses and commas: `(1, 2, 3)`. The comma is important. A tuple with just one item is `(5,)`; `(5)` is simply the number 5 in parentheses.

```python
# A tuple containing integers
numbers = (1, 2, 3, 4)

# A tuple containing different types of elements
mixed_data = (1, "Alice", 3.4, True)

print(type(numbers))
print(type((5,)))
```

```python
# You can access items using their position, just like with a list.
names = ("Alice", "Bob", "Charlie")
print(names[0])   # Alice: the first item is at position 0
print(names[-1])  # Charlie: -1 means the last item
```

```python
colors = ("red", "green")
# This will raise a TypeError:
# colors[0] = "blue"

# But this is okay: we are making a NEW tuple and giving it the same name.
colors = ("blue", "green")
print(colors)
```

```python
# You can use + to join two tuples. This creates a NEW tuple.
t1 = (1, 2)
t2 = (3, 4)
t3 = t1 + t2
print(t3) 
```

```python
# Use in to ask whether a value is present. The answer is True or False.
t = (1, 2, 3, 4)
print(2 in t)
print(9 in t)
```

One more useful trick: you can **unpack** a tuple into separate variables. The number of variables must match the number of items. We will use this again when we loop over dictionaries and when we use `zip`.

```python
product, price = ("Laptop", 1000)
print(product)
print(price)
```

# 💡 3. Comprehensions

## List comprehensions

Comprehensions are a cool little tool for building a new collection from something we can loop over. You may spend your whole programming life without using them once, but they can save you a lot of typing!

Let's first make a list of the numbers 1 through 10 using the for loop we already know. Remember: `range(1, 11)` starts at 1 and stops **before** 11.


```python
#create an empty list
my_list = []
#and start appending numbers
for k in range(1,11):
    my_list.append(k)
#show
my_list
```

We can do the same thing with a **list comprehension** in one line. Read the code below as: "make a list containing `k` for each `k` in `range(1, 11)`." The result is exactly the same list as above.

```python
my_list = [k for k in range(1,11)]
my_list
```

```python
#another example, makes a list with every number in 1,..,10 divided by 2
my_list = [k/2 for k in range(1,11)]
my_list
```

```python
#or their square!
my_list = [k**2 for k in range(1,11)]
my_list
```

Back to the list comprehension. The general shape is `[what_to_put_in_the_list for item in something_we_can_loop_over]`.
- The part after `for` tells Python where to get the items. Here, `range(1, 11)` gives us the numbers 1 through 10.
- The part before `for` tells Python what to put in the **new** list. We can keep `k`, divide it by 2, or square it.

Another cool thing we can do is add a condition at the end. Let's first do it with a regular loop, so we can see what the shorter version replaces.


```python
even_numbers = []
for k in range(0, 11):
    if k % 2 == 0:
        even_numbers.append(k)
even_numbers
```

Now for the same idea in one line. `k % 2` is the remainder after dividing `k` by 2, so `k % 2 == 0` keeps the even numbers. The `if` at the **end** filters: numbers that fail the test never enter the new list.

```python
my_list = [k for k in range(0,11) if k%2==0]
my_list
```

## Dictionary comprehensions

We can also build dictionaries with comprehensions. The idea is the same, but there are two changes:
- Use `{}` instead of `[]`, because we want a dictionary rather than a list.
- Give each item a **key** and a **value**, separated by `:`. In `{k: k**2 for k in range(0, 11)}`, the key is the number and the value is its square.

The second example keeps only the even keys. Notice that the `if` still goes at the end.

```python
{k:k**2 for k in range(0,11)}
```

```python
{k:k**2 for k in range(0,11) if k%2==0}
```

What about a comprehension inside another comprehension? The **inner** one below makes the squares first. The **outer** one takes each square and divides it by 2. It works, but it is a bit hard to read! For this particular example, `[k**2 / 2 for k in range(0, 11)]` is simpler.

```python
[k/2 for k in [k**2 for k in range(0,11)]]
```

### Note

The part after `in` can be any **iterable**: something Python can go through one item at a time. A string works too. Python will give us one character at a time from `'Apostolos'`.

```python
[k for k in 'Apostolos']
```

## Set comprehension with conditions

You can have set comprehensions as well. Use `{}`, but leave out the `key: value` pair that a dictionary needs. Sets keep only unique items, so repeated letters disappear. Sets do **not** promise an order, so your output may display the letters in a different order from mine.

```python
# set of unique letters in the word 'Apostolos'
{k for k in 'Apostolos'}
```

You can also add a condition. Here we keep only vowels. We use `lower()` twice: once to check whether a letter is a vowel, and once to put a lowercase letter in the set. That way uppercase `A` and lowercase `a` would count as the same vowel.

```python
# set of unique vowels in the word 'Apostolos'
word = 'Apostolos'
vowels = {char.lower() for char in word if char.lower() in 'aeiou'}
print(vowels)
```

There is another kind of `if` we can put **before** `for`. It does not filter items out. Instead, it chooses **what to put in the list** for each item. The list below therefore has ten labels: one for every number from 1 to 10. Compare that with the even-number example above, which kept only six numbers.

```python
# List of 'even' or 'odd' for numbers 1-10
result = ['even' if k%2==0 else 'odd' for k in range(1,11)]
print(result)
```

# 4. Dictionaries' .keys(), .values(), and .items()

We have already seen how dictionaries connect a **key** to a **value**. Here the product name is the key, and its price is the value.

There are three useful ways to loop over a dictionary:
- `.keys()` gives us the keys (product names). Writing `for product in products` does the same thing; `.keys()` just makes the intention more explicit.
- `.values()` gives us the values (prices). This is useful when we only need the prices.
- `.items()` gives us both together. Each item is a tuple like `("Laptop", 1000)`. Remember tuple unpacking? We can put the two parts into `product` and `price` as we loop!

```python
products = {
    "Laptop": 1000,
    "Smartphone": 500,
    "Headphones": 150,
    "Charger": 20
}

for product in products.keys():
    print(product)
```

```python
# We get prices here, but not the product names that go with them.
for price in products.values():
    print(price)
```

```python
# Each item is a (product, price) tuple, unpacked into two variables.
for product, price in products.items():
    print(f"The price of {product} is ${price}.")
```

# 🔄 5. Advanced iterators (OPTIONAL)

These next two tools are handy, but you do not need them to understand the rest of today's notebook. You can always use a regular loop instead.

`zip` pairs items from two iterables by their positions: the first name with the first score, the second name with the second score, and so on. Each pair is a tuple. We unpack that tuple into `name` and `score` in the loop below.

If the lists have different lengths, `zip` stops when the shorter one runs out. It will not invent a missing score! Also, a `zip` object is used up as you loop over it; create a new one if you want to loop again.

```python
names = ["Alice", "Bob", "Charlie"]
scores = [85, 92, 78]

print(list(zip(names, scores)))  # See the three (name, score) tuples.

zipped = zip(names, scores)
for name, score in zipped:
    print(f"{name}: {score}")
```

You can also use `zip` to "unzip" a collection of pairs. This time we start with `[(name, score), ...]` and want one tuple of names and one tuple of scores.

The `*` passes each pair to `zip` separately. Then the first pieces get grouped together, and so do the second pieces. You do not need to memorize this trick, but it helps to know what it is doing when you see it!

```python
zipped_data = [("Alice", 85), ("Bob", 92), ("Charlie", 78)]
names, scores = zip(*zipped_data)

print(names)  # Outputs: ('Alice', 'Bob', 'Charlie')
print(scores)  # Outputs: (85, 92, 78)
```

`enumerate` gives you both an item's position and the item itself. This is useful when you need the number **and** the name. It starts counting at 0, just like list indexing. If you only need the name, a plain `for name in names` loop is simpler.

```python
names = ["Alice", "Bob", "Charlie"]

for index, name in enumerate(names):
    print(f"{index}: {name}")

# Want to display 1, 2, 3 instead? The list itself is unchanged.
for number, name in enumerate(names, start=1):
    print(f"{number}: {name}")
```
