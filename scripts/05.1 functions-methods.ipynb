# 🛠 1. Methods

Everything in Python is an **object**. 

- Strings are objects, lists are objects, dictionaries are objects, and so on. Even number sare objects!
- While we have abstracted away the details of what an object is, we have been using objects all along. 
- One important thing to know about objects is that they have **attributes** and **methods**. 


In this notebook, we'll be focusing on methods. 

- Methods are functions that belong to objects.
- Every object of the same type will have access to the same methods. 
- We have used several methods so far.

Methods will generally perform certain actions on the object, and can also take arguments, just like a function does (more on that in a minute).

Methods generally have the following form

    object.method(arg1,arg2,....)
    
We will not dive deeper into objects here. Let's revisit some examples of methods that we have been using all along!

One useful distinction: `len(my_list)` is a function call, while `my_list.append(6)` is a method call. In the second case, the dot tells us that we are using a method that belongs to `my_list`.

```python
# lists are objects and have their own built-in methods
my_list = [1,2,3,4]
```

```python
# len() is a function: we give it the list as an argument
len(my_list)
```

```python
# one of these methods is the .append() method that we've used before!
my_list.append(6)
my_list.append(6)
my_list
```

```python
# or the .count() method that counts how many times a value appears in the list
my_list.count(4)
```

```python
my_list.count(6)
```

Notice that `.append()` changed `my_list` itself: the two 6s are now part of our list. But not every method changes the object! Let's see what happens with a string.

```python
# remember .upper() and .lower() methods for strings?
hi = "hello"
hi.upper()
```

We got `HELLO`. But did we change `hi`? Let's check, and then save the uppercase string in a new variable.

```python
print(hi)

hi_upper = hi.upper()
print(hi_upper)
```

So `hi` is still `hello`! `.upper()` **returns a new string**; it does not change the original string. If we wanted to replace `hi`, we would write `hi = hi.upper()`.

There is another difference worth knowing: `.append()` changes the list, but it does not return the updated list. Let's try saving its result.

```python
append_result = my_list.append(7)
print(my_list)
print(append_result)
```

The list now contains 7, but `append_result` is `None`. This is not an error! `None` is a Python value that represents the absence of a value.

So don't write `my_list = my_list.append(7)`: that would replace your list variable with `None`. We will see more about returned values below.

# 🌀 2. Functions


A function is, broadly, a useful device that groups together a number of statements so they can run more than once. Functions become increasingly important as our code grows and we want to perform the same thing many times in our code. 

Functions are pretty similar to mathematical functions. Let's see an example.

- Consider the function f(x) = 5\*x
- what that does is get an argument, and multiply it by the number 5
- so f(5) = 25, f(0) = 0, and f(30)=150.

Functions work similarly in Python. We have already used built-in functions like `len()` and `print()`. To write our own function, we must first define it with the **def** statement.

## 2.1 Defining and calling a function

```python
def my_function(arg1, arg2):
    #statement 1
    #...
    #statement n
    #return something
    pass
```

So let's take a look at the structure of our function.

- Function definitions begin with `def`, followed by the function's name. Pretty straightforward.
- The input names are enclosed in parentheses. These names are called **parameters**; the actual values we give the function when we call it are called **arguments**. A function can also have zero inputs!
- The colon starts the function's code block. The lines inside that block are indented, just like the lines inside a for loop or an if statement.
- Inside the function we can have one or more statements.
- Optionally, our function can `return` a value. We will see exactly what that means in a minute.

`pass` just means 'do nothing'. It lets us write a placeholder function without leaving its code block empty. In our actual examples, we'll replace it with code that does something.

Let's see some examples.

```python
# a function that prints hello world
def hello():
    print('hello world')
```

Defining a function does not run it! We are simply telling Python what to do when we call it. That's why the cell above did not print anything.

To actually run it, we need `hello()`. The parentheses matter: they tell Python to call the function.

```python
hello()
```

This function had zero arguments and just did the same thing every time. We can call it again whenever we want!

## 2.2 Giving a function inputs

Let's now give our function a name to work with.

## 2.3 Returning a result

So far, our functions printed a message. Let's now write a function that gives a result back to the code that called it.

```python
def multiply(x, y):
    return x * y
```

```python
multiply(3,4)
```

```python
multiply(5,10)
```

```python
result = multiply(5.5,5)
```

```python
result
```

This function takes two arguments and multiplies them. We can also save the result to a variable, since the function **returns** that value!

### Printing is not the same as returning!

`print()` shows something on the screen. `return` gives a value back to the code that called the function, so we can save it or use it in another calculation.

Let's save the result of our welcome function and see what we get.

```python
result = welcome('Apostolos')
print('Saved result:', result)
```

The welcome message appeared, but the saved result is `None`. Our function printed a message; it did not return that message. A function without an explicit return value returns `None`.

Now let's do the same thing with `multiply()`.

```python
result = multiply(3,4)
print('Saved result:', result)
print('Double the result:', result * 2)
```

This time, we got 12 back, and we can use it to calculate 24. That's the point of returning a result!

Also, once Python reaches `return`, that function call ends. Any statements below that `return` in the function will not run during that call.

One notebook detail: Jupyter also displays the value of the last expression in a cell. That's why `multiply(3,4)` showed 12 even though our function did not use `print()`. Seeing output does not necessarily mean the function printed it!

## 2.4 Returning other kinds of values

The value that a function returns need not only be a number. For example, the following function returns a list!

```python
def calculate_values(x, y):
    return [x*x, y*y, x*y]
```

```python
calculate_values(5,10)
```

This list contains the square of x, the square of y, and their product.

As always, functions can contain more than one statement! Here is the same calculation, written step by step. We give this version a different name, so we can keep using both versions.

If we define a function again with the same name, Python replaces the old definition. So rerunning an earlier cell can change which version we are using. Different names help us avoid that confusion here.

```python
def calculate_values_step_by_step(x, y):
    square_x = x*x
    square_y = y*y
    product = x*y
    values = [square_x, square_y, product]
    return values
```

```python
calculate_values_step_by_step(5,10)
```

## 2.5 Using functions to simplify our code

Alright, imagine we have hundreds of titles to clean. Instead of writing the same cleaning code every time, we can put it in a function and reuse it!

Let's say we want to remove whitespace from the beginning and end of a title, and then make it lowercase.

```python
def clean_text(text):
    return text.strip().lower()
```

```python
print(clean_text('  Web Analytics  '))
print(clean_text('  Python  '))
```

Here, `.strip()` returns a string without the whitespace at its ends, and `.lower()` makes that result lowercase. Our function returns the cleaned string. It does not change the original string.

We can also call our function from a for loop. Nothing new about the loop: we just use our function for each title!

```python
titles = ['  Web Analytics  ', '  Python  ', '  Beautiful Soup  ']
cleaned_titles = []

for title in titles:
    cleaned_titles.append(clean_text(title))

print(cleaned_titles)
```

Same function, three different inputs. If we later change how we want to clean titles, we can update the function and rerun the cells that use it. We don't have to hunt down the same cleaning code in several places!

This will be useful when scraping too. We can put a repeated task, like extracting information from one article, into a function and call it for every article.

## 2.6 Optional: functions within functions

You can also define a function inside another function. The example below is a little more involved, so don't worry if it takes a while to follow it! You can skip it for now and come back to it later.

Here, `calculate_statistics()` takes a non-empty list of numbers. Inside it, we define three smaller functions:

- `mean()` returns the average: the sum of the numbers divided by how many there are.
- `median()` sorts the numbers and finds the middle. If there are two middle numbers, it returns their average. `//` gives us an integer index.
- `mode()` returns a list of the numbers that appear most often. It uses `Counter`, from Python's standard library, to count how many times each number appears. No extra package to install! If several numbers tie, we return all of them.

These three functions are defined inside `calculate_statistics()`, and that is where we call them. At the end, we collect their returned values in a dictionary.

```python
# you can -- of course -- define functions within functions!
def calculate_statistics(nums):
    def mean(nums):
        return sum(nums) / len(nums)
    
    def median(nums):
        sorted_nums = sorted(nums)
        mid = len(nums) // 2
        if len(nums) % 2 == 0:
            return (sorted_nums[mid - 1] + sorted_nums[mid]) / 2
        else:
            return sorted_nums[mid]
    
    def mode(nums):
        from collections import Counter
        counts = Counter(nums)
        max_count = max(counts.values())
        return [num for num, count in counts.items() if count == max_count]

    return {
        'mean': mean(nums),
        'median': median(nums),
        'mode': mode(nums)
    }


numbers = [4, 5, 5, 6, 7, 8, 8, 9]
stats = calculate_statistics(numbers)
print(stats)
```

For our list, the mean and median are both 6.5. The mode is `[5, 8]`, because both numbers appear twice.

The main idea is still the same: a function takes inputs, does some work, and can return a result. Here, the result happens to be a dictionary!
