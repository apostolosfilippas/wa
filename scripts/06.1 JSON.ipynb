# 🚪 1. Introduction

JSON (JavaScript Object Notation) is widely used data format. It originated from JavaScript, but it's now used nearly in every programming language. JSON is widely used because it has many advantages:
- **Interpretability**: It is easy for humans to read and write, and it's easy for machines to parse and generate.
- **Standardized**: There is a standard JSON format that is widely adopted.
- **Lightweight**: JSON is great for sending and receiving data through APIs.
- **Structured**: JSON uses variable_name-variable_value pairs, similar to Python dictionaries.


JSON files have some simple syntax rules:
- Data is in name/value pairs (like Python dictionaries).
- Data is separated by commas.
- Curly braces {} hold objects (key-value pairs).
- Square brackets [] hold arrays (lists).

JSON can represent the following data types:
- String
- Number
- Object (JSON object)
- Array
- Boolean (true or false)
- null


````python
import json
````

# 2. JSON in Python
Python has a built-in module called json which can be used to work with JSON data.

#### Parsing JSON strings to Python objects

The `json.loads()` method can be used to parse a JSON string into a Python object.
- The Python object you get depends on the structure of your JSON string. 

Here are two examples:

````python
# some JSON data:
json_data = '{"name": "Apostolos", "last_name": "Filippas", "city": "New York"}'

# parse JSON:
data = json.loads(json_data)

# the result is a Python dictionary:
print(f"The type of the `data` object is {type(data)}")  
print(data)

# access the data:
print(f"The name is {data.get('name')}")
````

````python
# Example 1
json_data = '["Chicago","New York","Los Angeles"]'
data = json.loads(json_data) # parse JSON
print(f"The parsed object is {data.__class__.__name__}.")  
print(data) 
````

````python
json_data = '{"name": "Apostolos", "last_name": "Filippas", "city": "New York"}'
data = json.loads(json_data) # parse JSON
print(f"The parsed object is {data.__class__.__name__}.")  
print(data) 
````

#### Writing Python objects to JSON strings

The reverse of `loads()` is `dumps()`. 
- `dumps()` takes a Python object and returns a string representation of it.

Here are two examples:

````python
# EXAMPLE 1 - dumping a Python dictionary:
python_data = {
  "name": "Apostolos",
  "last_name": "Filippas",
  "city": "New York",
  "married": False,
  "children": None
}

# convert into JSON string
json_data = json.dumps(python_data)
print(f"The type of the `json_data` object is {type(json_data)}")
print(f"The value of the `json_data` object is {json_data}")
````

````python
#EXAMPLE 2 - dumping a Python list:
python_data = [
    {"name": "John", "age": 30, "city": "New York", "married": True, "children": None},
    {"name": "Peter", "age": 45, "city": "Chicago", "married": False, "children": ("Ann","Betty")},
]

# convert into JSON string
json_data = json.dumps(python_data)
print(f"The type of the `json_data` object is {type(json_data)}")
print(f"The value of the `json_data` object is {json_data}")
````

#### Python/JSON data type mapping
The mapping between Python and JSON data types is relatively straightforward, but there are some important gotchas to be aware of.

Below you can find the mapping between Python and JSON data types:

|  Python  |  JSON   |
|:--------:|:-------:|
|  `dict`  | Object  |
|  `list`  | Array   |
|  `tuple` | Array   |
|  `str`   | String  |
|  `int`   | Number  |
|  `float` | Number  |
|  `True`  | `true`  |
|  `False` | `false` |
|  `None`  | `null`  |


Note that in the previous examples
- "married", which is "True" or "False" in Python, gets translated to a lowercase true or false in the JSON string. 
- "children" is None in the dictionary of the first example, but in the JSON data, it is represented by "null".
- "children" is a tuple in the dictionary of the second example, but in the JSON data, it is represented by an array. JSON does not have a tuple data type.


#### Writing and reading to files

We can also write and read JSON data to/from files.
To do so, you can first use `dumps()` to convert a Python object to a JSON string, and then use `write()` to write to a file as a string. Similarly, you can use `read()` to read a JSON string from a file, and then use `loads()` to convert it to a Python object.

However, you can also use `dump()` and `load()` to write and read JSON data to/from files directly.
Here's an example:

````python
from pathlib import Path

# let's save this file in the course's temp folder
# notebooks can start in wa, scripts, or homework, so let's check where we are
course_folder = Path.cwd().resolve()
if course_folder.name in {'scripts', 'homework'}:
    course_folder = course_folder.parent
if not (course_folder / 'pyproject.toml').is_file():
    raise RuntimeError('Open the wa course folder in VS Code, then restart this notebook.')

output_folder = course_folder / 'temp'
output_folder.mkdir(exist_ok=True)
json_file = output_folder / 'example.json'

person = {
    "name": "Bob",
    "age": 25,
    "city": "Funtown"
}

with open(json_file, 'w', encoding='utf-8') as file:
    json.dump(person, file)

print(f'Saved our file to {json_file}')
````

````python
# you can make the output more readable by using the indent argument
# the following will format the JSON output with an indentation of 4 spaces for each level
with open(json_file, 'w', encoding='utf-8') as file:
    json.dump(person, file, indent=4)

# and now let's read it back into Python!
with open(json_file, 'r', encoding='utf-8') as file:
    saved_person = json.load(file)

print(saved_person)
print(type(saved_person))
assert saved_person == person
````

#### Handling errors
JSON, being a text-based format, can sometimes contain errors, either due to human editing or other factors. Python’s json module will raise exceptions when encountering improperly formatted JSON.

A common exception is  `json.JSONDecodeError` 
- This exception is raised when there's an error decoding a JSON document. 
- It's the most common error you'll encounter when parsing JSON in Python.

To gracefully handle errors, use the `try` and `except` blocks.

````python
malformed_json = "{ 'name': 'Alice' }"  # Single quotes are not valid in JSON

try:
    data = json.loads(malformed_json)
except json.JSONDecodeError:
    print("Failed to decode JSON")
    data = {}
````

Tips for Debugging:
- Ensure that the JSON strings use double quotes for keys and string values.
- Validate the JSON using online tools to pinpoint errors.

# 4. JSON Lines (JSONL)

Sometimes we want to save a lot of records, one after another -- e.g., page views on a website, or reviews we are collecting as we scrape. Each review is a JSON object, but how do we store many of them in one file?

The solution is to use a `.jsonl`  (JSON Lines) file. This file has one valid JSON value on each line.

Let's compare the two formats using some made-up website activity.

**A regular JSON file:** one list containing all our records.

```json
[
  {"page": "/home", "event": "page_view", "user_id": 101},
  {"page": "/courses", "event": "page_view", "user_id": 102},
  {"page": "/contact", "event": "click", "user_id": 101}
]
```

**A JSONL file:** the same records, but each on its own line.

```text
{"page": "/home", "event": "page_view", "user_id": 101}
{"page": "/courses", "event": "page_view", "user_id": 102}
{"page": "/contact", "event": "click", "user_id": 101}
```

Notice: no outer square brackets, and no commas **between** records.
- In the first example, the entire file is a JSON array
- In the second example, each line is a separate JSON object


#### Writing a JSONL file

We don't need another package for this. We can use the `json` module we already imported.

Let's write each dictionary as a JSON string, followed by a newline (`\n`). We'll use the `temp` folder we created earlier, so run the notebook from the top first.

Don't use `indent=4` here: that would spread a record over several lines. We want one complete record per line. Use UTF-8, just as we did for our JSON file.


````python
# some made-up website activity
events = [
    {"page": "/home", "event": "page_view", "user_id": 101},
    {"page": "/courses", "event": "page_view", "user_id": 102},
    {"page": "/contact", "event": "click", "user_id": 101}
]

jsonl_file = output_folder / 'events.jsonl'

# 'w' writes a fresh file, replacing the previous example if it exists
with open(jsonl_file, 'w', encoding='utf-8') as file:
    for event in events:
        file.write(json.dumps(event, ensure_ascii=False) + '\n')

print(f'Saved our records to {jsonl_file}')

# let's look at what we actually wrote
with open(jsonl_file, 'r', encoding='utf-8') as file:
    print(file.read())
````

#### Reading a JSONL file

For a regular JSON file, we used `json.load(file)` to read the whole document. For JSONL, we read the file **line by line** and use `json.loads(line)` on each line.

Remember the difference: `load()` reads a file; `loads()` parses a string. Here, each `line` is a string containing one record.

If we try `json.load()` on this entire JSONL file, we'll get a `JSONDecodeError` about extra data: there is more than one JSON value. [Python's JSON documentation](https://docs.python.org/3/library/json.html).


````python
loaded_events = []

with open(jsonl_file, 'r', encoding='utf-8') as file:
    for line in file:
        event = json.loads(line)
        loaded_events.append(event)

print(loaded_events)
print(f'We read {len(loaded_events)} records')
print(type(loaded_events[0]))
assert loaded_events == events
````

#### Working with one record at a time

The previous example put every record into a list. That's useful, but it still keeps all the records in memory!

If the file is very large, we can instead do something with each record and then move on to the next one. Let's count the page views without building a big list.


````python
page_views = 0

with open(jsonl_file, 'r', encoding='utf-8') as file:
    for line in file:
        event = json.loads(line)
        if event['event'] == 'page_view':
            page_views += 1

print(f'There were {page_views} page views')
````

#### Adding another record

What if another page view comes in? We can open the file in **append mode**, `'a'`, and add one more line without rewriting the earlier records. Our writing cell ends every record with a newline, so the new record starts on a fresh line.

Be careful: `'w'` replaces the file; `'a'` adds to it. Running the cell below twice adds the same example twice. To reset our example, run the writing cell again first.


````python
new_event = {"page": "/courses", "event": "page_view", "user_id": 103}

with open(jsonl_file, 'a', encoding='utf-8') as file:
    file.write(json.dumps(new_event, ensure_ascii=False) + '\n')

# now let's see all four records
with open(jsonl_file, 'r', encoding='utf-8') as file:
    for line in file:
        print(json.loads(line))
````

#### A few things to watch out for

- Every line must be valid JSON. Blank lines are not valid JSONL records.
- Use `json.dumps()`, not `str()`, to turn a dictionary into JSON. Python's single quotes, `True`, and `None` are not JSON syntax.
- A newline **inside a string value** is fine: `json.dumps()` escapes it for us. That is different from putting a physical newline in the middle of a record.
- If you get a `JSONDecodeError`, check the line you are trying to read. Don't silently throw it away-- we might lose a record we need!

That's it! Same JSON tools, just a different way of organizing the file.
