## 4. Navigating around a website with Selenium 

````python
# initialize our browser
browser = webdriver.Chrome()
browser.set_page_load_timeout(40)
browser.maximize_window()
wait = WebDriverWait(browser, 30, ignored_exceptions=(StaleElementReferenceException,))

# let's go to the page of Bklyn House Hotel on Booking.com
link = "https://www.booking.com/hotel/us/bklyn-house-new-york-brooklyn.html"
try:
    browser.get(link)
    wait.until(EC.presence_of_element_located((By.TAG_NAME, 'body')))
except Exception:
    browser.quit()
    raise
````

## 4.1 XML
The browser still uses HTML for both static and dynamic content. Before we learn XPath, let's briefly look at __**XML**__ (Extensible Markup Language), another format for structured data.

- XML is a markup language,designed to store, structure, and transport data or information.
- It focuses on representing the content of data rather than specifying how it should be displayed (unlike HTML)
- XML is widely used for data interchange and storage, providing a standardized way to format and organize data.
- XML documents use tags to enclose data elements, creating a hierarchical structure. 


Here's an XML example:

```xml
<bookstore>
  <book>
    <title lang="en">Harry Potter and the mid-life crisis</title>
    <description>Go on a journey with Harry Potter and his friends as they navigate the challenges of middle age by buying a Porsche.</description>
    <price>29.99</price>
  </book>
</bookstore>
```

In a nutshell:

| Feature          | **HTML (HyperText Markup Language)**                 | **XML (eXtensible Markup Language)**                     |
| ---------------- | ---------------------------------------------------- | -------------------------------------------------------- |
| **Goal**         | Display data — defines how data looks on a web page. | Store and transport data — defines what data *is*.       |
| **Design focus** | Presentation (how information is shown).             | Structure and meaning (what the information represents). |
| **Tags**         | HTML uses predefined tags like `<p>`, `<div>`, ... that have meaning built into browsers.   | XML has no predefined tags — you invent your own, like `<book>` or `<price>`. XML is extensible — that’s what the “X” stands for. |
| **Syntax**       | HTML is lenient — browsers can interpret sloppy or missing tags (<p><b>Hi will still render).   | XML is strict — every tag must close, nesting must be perfect, and case matters (`<Tag>` is not the same as `<tag>`).|

## 4.2 XPath

XPath lets you “walk” the XML tree, and select or extract specific nodes.
- XPath can be used to locate and interact with elements both within XML or HTML documents, making it particularly useful for tasks like web scraping and automated testing.
- You can find the XPATH by right click the web page, click the inspect button and select the element you want to check. It's very similar to access content with id, class and other attributes, but gives us even more freedom.

A sample XPath is shown below: 



<br>
<img src="https://raw.githubusercontent.com/apostolosfilippas/wa/refs/heads/main/assets/selenium-7.webp" width="800" height="280">
<br>

- `//`:  This selects all elements in the document that match the criteria that follow, regardless of their location within the document's hierarchy.
- `tagname`: This selects all elements with the specified tag name.
- `[@attribute='value']`: This selects all elements that have the specified attribute with a value equal to the specified value.

You do not have to remember everything, but here's some basic syntax:

| Example                       | Meaning                                                                  |
| ----------------------------- | ------------------------------------------------------------------------ |
| `/bookstore/book/title`       | Selects all `<title>` elements inside `<book>` inside `<bookstore>`.     |
| `//title`                     | Selects **all** `<title>` elements anywhere in the document.             |
| `//book[@category="fiction"]` | Selects `<book>` elements whose `category` attribute equals `"fiction"`. |
| `//book/title`         | Selects the `<title>` elements inside all `<book>` elements.                    |
| `(//book)[1]`                 | Selects the *first* `<book>` element.                                    |

In Selenium, our XPath needs to select an **element**. To get its text, use the element's `.text` property; `text()` in XPath selects a text node, not a browser element.


## Manipulating content using XPath 

Let's try to go to the reviews tab.



<br>
<img src="https://raw.githubusercontent.com/apostolosfilippas/wa/refs/heads/main/assets/selenium-8.png" >
<br>

To find the Guest reviews tab, we can use its `data-testid` attribute:

`[data-testid="Property-Header-Nav-Tab-Trigger-reviews"]`

Or, in XPath:

`//a[@data-testid='Property-Header-Nav-Tab-Trigger-reviews']`

Notice that we are looking for the actual link, not a span somewhere inside it.

We also need to wait for the reviews to load. `time.sleep(3)` just waits three seconds, whether the page is ready or not. `WebDriverWait` waits for something specific, up to a time limit. [More on waits](https://www.selenium.dev/documentation/webdriver/waits/).


````python
# a small helper: wait for a button/link, scroll to it, and click it
# pages sometimes replace their elements while loading, so we find it again if needed
def click_when_ready(by, selector):
    def click_element(browser):
        try:
            element = browser.find_element(by, selector)
            if not element.is_displayed() or not element.is_enabled():
                return False
            browser.execute_script("arguments[0].scrollIntoView({block:'center'});", element)
            element.click()
            return True
        except (NoSuchElementException, StaleElementReferenceException, ElementClickInterceptedException):
            return False

    wait.until(click_element)


def open_reviews():
    xp = "//a[@data-testid='Property-Header-Nav-Tab-Trigger-reviews']"
    try:
        click_when_ready(By.XPATH, xp)
        wait.until(EC.visibility_of_all_elements_located((By.CSS_SELECTOR, '[data-testid="review-card"]')))
    except TimeoutException as error:
        browser.quit()
        raise RuntimeError(
            'The reviews did not load. Booking may be showing a browser check, '
            'no reviews, or a different layout. Check the page in regular Chrome '
            'and email the TA with a screenshot. We have not scraped any reviews.'
        ) from error

open_reviews()
print('The reviews are open!')
````

## Let's click "Continue reading"

Some hotel responses are shortened on the screen. Let's expand one and see what changes.

These are the hotel's responses, not the guests' reviews. Not every response needs expanding, so we'll check whether a button is there first. Once expanded, this button disappears-- there isn't a "Show less" button to click here.


````python
toggle_css = 'button[data-testid="review-pr-toggle"]'
toggles = browser.find_elements(By.CSS_SELECTOR, toggle_css)

if toggles:
    before = len(toggles)
    click_when_ready(By.XPATH, '(//button[@data-testid="review-pr-toggle"])[1]')

    # the button disappears after expanding the response; find the remaining ones again
    wait.until(lambda browser: len(browser.find_elements(By.CSS_SELECTOR, toggle_css)) < before)
    print('Expanded one hotel response!')
    print('Remaining buttons:', len(browser.find_elements(By.CSS_SELECTOR, toggle_css)))
else:
    print('No shortened hotel responses on this page. Nothing to expand!')
````

## Identifying elements through their full XPath


An absolute XPath starts at the root of the document, and follows the full path to an element. For example:

`/html/body/div[1]/div[2]/button`

What do `div[1]` and `div[2]` mean? Let's break it down:
- `/html/body`: start at the page's body.
- `div[1]`: go to the **first div directly inside the body**.
- `div[2]`: now, inside that first div, go to its **second div child**.
- `button`: look for buttons directly inside that second div.

So the page might look like this:

```text
body
├── div             ← div[1] under body
│   ├── div         ← first div inside that div
│   └── div         ← div[2] inside that div
│       └── button  ← this is what we're looking for!
└── div             ← second div under body; NOT the one we mean
```

Notice two things: **XPath counts from 1, not 0** like Python lists. Also, these numbers count only div children of the current parent -- not every child, and not every div on the whole page. If a paragraph comes before a div, it doesn't change that div's number.

You can copy one by right-clicking an element in Chrome's Inspect panel and choosing **Copy → Copy full XPath**.

The problem: if the website adds a div or moves the button, the path changes. A selector based on a meaningful attribute, like `data-testid`, is usually easier to maintain. Don't expect an old copied path to work forever!


## Dealing with multiple elements with an XPath

There are multiple buttons on the page. Let's find all of them and inspect what they contain.

Here is an XPath that matches buttons on the page:

//button[@type='button']

````python
# Grabbing all the buttons on a page
xpath_buttons = "//button[@type='button']"

time.sleep(2)
elements = browser.find_elements(By.XPATH, xpath_buttons)

print(type(elements))
print(len(elements))
elements[:10]
````

**How do we know which element is which??**
- We can extract the `innerHTML` or `property` or `text` using the by `.get_attribute('innerHTML')`, `.get_property()` or `.text` methods respectively

````python
# let's look at the HTML of a few buttons, rather than print the whole page
for element in elements[:5]:
    print(element.get_attribute('outerHTML')[:500])
    print()
````

Is there **another** way of identifying the button?

````python
# another way: look only for buttons that expand hotel responses
response_buttons = browser.find_elements(By.XPATH, "//button[@data-testid='review-pr-toggle']")
for i, button in enumerate(response_buttons):
    print(i, button.text)
````

## 5. Using the CSS selector to find elements

We can use CSS Selectors instead of XPath to find and interact with elements

##### **What is a CSS Selector?**

A CSS selector is like a set of instructions that tells a web browser how to find and style elements on a web page. Remember that elements have different attributes; we can use CSS to look for elements with specific attribute values.


##### **Examples of CSS Selectors:**
1. Element Selector

    < button > or < a > will point to elements with those respective types

2. Class Selector --> .
    
    '.expand' points to all elements with class='expand'

3. ID Selector--> #
    
     #submit-button points to the element with id="submit-button"

4. Attribute Selector --> [type=" "] [attribute="value"]

    eg. [ type="text" ] points to all elements with type="text"

````python
# locating by CSS selector
example_button = 'button[data-testid="review-pr-toggle"]'
response_buttons = browser.find_elements(By.CSS_SELECTOR, example_button)
print(f'Found {len(response_buttons)} response buttons')
if response_buttons:
    print(response_buttons[0].text)
````

**How did this work?**

`button`: select button elements.

`[data-testid="review-pr-toggle"]`: keep only elements with this exact attribute value.

Put them together, and we get the buttons that expand hotel responses. We don't want to click *every* button on the page-- some do completely different things!


There are multiple ways to locate elements using Selenium. For the full list, you can refer to [the Selenium documentation](https://www.selenium.dev/documentation/webdriver/elements/locators/).


## Trying to find a button that doesn't exist

What if we look for an element that isn't on the page? We'll get a `NoSuchElementException`.

Let's do that on purpose, but catch the error so the rest of our notebook can still run.


````python
try:
    browser.find_element(By.CSS_SELECTOR, '#this-button-does-not-exist')
except NoSuchElementException:
    print("NoSuchElementException: that button doesn't exist!")
````

````python
# a helper function to check if the button exists on the page
def check_exists_by_css(selector):
    try:
        browser.find_element(By.CSS_SELECTOR,selector)
    except NoSuchElementException:
        return False
    return True

def check_exists_by_xpath(xp):
    try:
        browser.find_element(By.XPATH,xp)
    except NoSuchElementException:
        return False
    return True
````

````python
# so if we try to find a nonsense element, it will just return False instead of breaking the code

check_exists_by_css('hahaha')
````

# Expanding hotel responses

````python
# let's expand all shortened hotel responses on this page
continue_reading_xp = "//button[@data-testid='review-pr-toggle' and contains(., 'Continue reading')]"
expanded = 0

# find the remaining buttons again after each click: the page can replace them
while browser.find_elements(By.XPATH, continue_reading_xp):
    before = len(browser.find_elements(By.XPATH, continue_reading_xp))
    click_when_ready(By.XPATH, continue_reading_xp)
    wait.until(lambda browser: len(browser.find_elements(By.XPATH, continue_reading_xp)) < before)
    expanded += 1

print(f'Expanded {expanded} hotel responses')
````
