# 🐼 2. The Pandas Package

The Pandas package gives us the **DATAFRAME** --- one of the main data structures used in data analytics.

- A dataframe is an Excel sheet on steroids.
- More scientifically, a dataframe is a 2-dimensional "labeled" data structure: it has rows and columns, and different columns can store different types of data.

We can create dataframes from CSV files, Excel spreadsheets, SQL databases, JSON, webpages, and even data already stored in our Python code.

First, let's use the [Auto MPG dataset from UCI](https://archive.ics.uci.edu/dataset/9/auto+mpg). It contains information about older cars and their fuel consumption. `mpg` means miles per gallon; a higher number means the car goes farther on a gallon of fuel.

A copy is already in our `data` folder, so we don't need to download anything to run this notebook.

````python
column_names = ['mpg', 'cylinders', 'displacement',
                'horsepower', 'weight', 'acceleration',
                'model_year', 'origin', 'car_name']

# this file uses spaces between values, rather than commas
# sep=r'\s+' tells pandas to treat one or more spaces as a separator
# header=None means the file has no row of column names; we supply our own
mpg_df = pd.read_csv(data_folder / 'auto-mpg.data-original',
                     sep=r'\s+',
                     header=None,
                     names=column_names)
````

We now have the data loaded in a pandas dataframe. Let's see some of the (MANY!) ways pandas makes it convenient to explore a dataset.

## Selecting columns

We saw how to select columns, but here it is again. One column name gives us a **Series** (one labeled column); a list of column names gives us another **DataFrame**.

````python
# selecting columns is done using the `[]` operator,
# which accepts one column name or a list of several names
mpg_df[["cylinders", "car_name"]].head(5)
````

````python
# the dot notation is convenient when a column name is a valid Python name
# [] also works for names with spaces, and is the safer choice for adding columns
mpg_df.car_name.head(5)
````

## Selecting rows

There are two options for selecting rows in pandas:

- `.loc`: select rows using their **labels**.
- `.iloc`: select rows using their **positions**, starting at 0.

Right now, the labels and positions happen to be the same. That won't always be true! We can use a date, a user ID, or another column as our row labels.

There is also a small difference when slicing: `.iloc[:5]` stops BEFORE position 5, while `.loc[:5]` includes label 5. Let's see it.

````python
# get a single row
mpg_df.iloc[5]
````

````python
# selecting the first 5 rows
mpg_df.iloc[:5]
````

````python
# .loc includes label 5, so this gives us SIX rows: labels 0 through 5
mpg_df.loc[:5]
````

One can also select rows that match a particular condition. Say I want to only see cars with an acceleration time of less than 10 seconds.

The condition inside the brackets gives us True or False for each row. Pandas keeps the rows where it is True.

````python
mpg_df[mpg_df["acceleration"] < 10].head(5)
````

If we use car names as our row labels, we can select cars by name with `.loc`.

A label doesn't have to be unique: if several rows have the same car name, `.loc` can return all of them.

````python
# this creates a new dataframe; it does not change mpg_df
car_index_df = mpg_df.set_index('car_name')
car_index_df.loc[['amc rebel sst', 'plymouth fury iii']]
````

## Creating a dataframe

Often, one wants to create a data frame from information that is available "in code." This information might be results of prior computations that aren't already in pandas, or maybe just some small static dataframe that stores some info. 

There are two common ways to do this: 
- lists-of-lists with an additional list of column names
- lists of dictionaries 

I prefer the latter since the data in this case is self-descriptive, order isn't important, and missing data is handled more smoothly, but I'll give examples below for both. 

````python
# list-of-lists approach

list1 = ['studentA',22,'(929)-000-000']
list2 = ['studentB',np.nan,'(646)-000-000']
list3 = ['studentC',30,'(917)-000-000']
list4 = ['studentD',31,'(646)-001-001']
list5 = ['studentE',np.nan,'(929)-001-001']
list6 = ['studentF',30,'(917)-001-001']
list7 = ['studentG',30,'(470)-001-001']

list_of_lists = [list1, list2, list3, list4, list5, list6, list7]
column_names = ['Name','Age','Mobile']

lol_df = pd.DataFrame(list_of_lists,columns=column_names)
lol_df.head(5)

````

````python
# this is the list of dicts approach
alice = {"name": "alice", "age": 5, "mobile":"555-222-9000"}
bob = {"name": "bob", "age": 100}
casey = {"age":35, "name": "casey", "mobile":"1-877-kars-4kids"}

list_of_dicts = [alice, bob, casey]
lod_df = pd.DataFrame(list_of_dicts)
lod_df.head(5)
````

We can also add columns ( they should have the same number of rows as the dataframe they are being added to )

````python

lol_df['Business Major'] = ['yes','no','yes','yes','yes','no','yes']
lol_df['Years Experience'] = [1,4,2,6,0,3,0]

lol_df.head(5)
````

## Operations with columns

What about operations on entire columns? This can make data munging much easier!

Let's take the difference between age and years of experience:


````python

lol_df["Age"] - lol_df["Years Experience"]
````

Notice the missing values, which pandas displays as `NaN`. Missing does NOT mean zero! It means we don't know the value.

Pandas has [methods for dealing with missing data](https://pandas.pydata.org/pandas-docs/stable/user_guide/missing_data.html). What we should do depends on the question we're trying to answer.

Let's (for some reason) fill the missing ages with the average age. This is just an example, not a rule that we should use for every missing value!

````python
# fillna returns a new Series here; it does not change the original Age column
lol_df['Age'].fillna(lol_df['Age'].mean()) - lol_df['Years Experience']
````

# 🏋️‍♀️ ✅. Solutions: Employee Reviews Analysis

Here is one way to do it. If your code looks different but gives the same result, that's fine!

Remember: this is the fictional teaching sample in `data/employee_reviews.csv`, not actual employee-review data.

## 1. Setup and Data Loading

````python
# parse_dates tells pandas that Date contains dates, not ordinary text
reviews_df = pd.read_csv(data_folder / 'employee_reviews.csv',
                         parse_dates=['Date'])
````

## 2. Exploration and Cleaning

Let's take a peek, check the column types, and count missing values and duplicates.

````python
reviews_df.head()
````

````python
# info() prints its own summary; we don't need print() around it
reviews_df.info()
````

````python
print('Missing values in each column:')
print(reviews_df.isna().sum())
print('Repeated rows:', reviews_df.duplicated().sum())
````

## 3. Data Analysis

`.value_counts()` counts rows in each category. `.groupby()` puts rows into groups, so we can calculate something for each group.

For example, `groupby('Role')['Overall_rating'].mean()` takes the mean of the reviews within each role. One average per role, without writing a loop!

````python
# how many reviews are there for each job type?
job_type_counts = reviews_df['Job_type'].value_counts()
print(job_type_counts)

# mean rating within each role, then sort from highest to lowest
top_roles = reviews_df.groupby('Role')['Overall_rating'].mean().sort_values(ascending=False).head(5)
display(top_roles)

avg_rating_per_department = reviews_df.groupby('Department')['Overall_rating'].mean()
display(avg_rating_per_department)
````

````python
# ratings are whole numbers, so give each rating its own bin
sns.histplot(reviews_df['Overall_rating'], discrete=True, color='#66c2a5')
plt.title('Overall Ratings — Fictional Employee Reviews')
plt.xlabel('Overall Rating (1–5)')
plt.ylabel('Number of Reviews')
plt.xticks(range(1, 6))
plt.ylim(bottom=0)
plt.show()
````

````python
# several reviews can have the same pair of ratings, so dots may overlap
sns.scatterplot(data=reviews_df, x='work_life_balance', y='work_satisfaction',
                alpha=0.6, color='#66c2a5')
plt.title('Work-Life Balance vs. Work Satisfaction — Fictional Reviews')
plt.xlabel('Work-Life Balance Rating (1–5)')
plt.ylabel('Work Satisfaction Rating (1–5)')
plt.xticks(range(1, 6))
plt.yticks(range(1, 6))
plt.show()
````

## 4. Indexing and Selection

Let's use job roles as row labels in a NEW dataframe. The original `reviews_df` keeps its `Role` column, so we can run this cell again without losing it.

````python
reviews_by_role = reviews_df.set_index('Role')

# a role can appear more than once; .loc selects the matching rows
sde_reviews = reviews_by_role.loc[['Software Development Engineer']]
display(sde_reviews)

# the 50th row is at position 49, not 50
fiftieth_review = reviews_by_role.iloc[49]
display(fiftieth_review)
````

## 5. Bonus (Advanced)

Let's keep reviews with unknown locations, select roles starting with 'S', and put our months in date order.

````python
# copy() lets us fill missing locations without changing reviews_df
reviews_with_places = reviews_df.copy()
reviews_with_places['Place'] = reviews_with_places['Place'].fillna('Not provided')

top_places = reviews_with_places.groupby('Place')['Overall_rating'].mean().sort_values(ascending=False).head(5)
display(top_places)
````

````python
# na=False means a missing role would not match our condition
s_starting_roles = reviews_df[reviews_df['Role'].str.startswith('S', na=False)]
s_starting_roles.head(10)
````

````python
# .dt lets us use date-specific methods on a column of dates
# to_period('M') keeps the year AND month: 2025-01, 2025-02, and so on
review_months = reviews_df['Date'].dt.to_period('M')

# sort_index() puts these year-months in date order, not alphabetical order
monthly_reviews = review_months.value_counts().sort_index()

monthly_reviews.plot(kind='bar', color='#66c2a5')
plt.title('Reviews per Month — Fictional Employee Reviews')
plt.xlabel('Year-Month')
plt.ylabel('Number of Reviews')
plt.ylim(bottom=0)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
````
