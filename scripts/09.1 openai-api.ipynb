# 4. Using OpenAI's API

Now we'll make a request to OpenAI's API! But first, we need to retrieve our API key.

Select the course's `.venv` Python environment as your notebook kernel, then run the cells in order.

## 4.1 Retrieve API key

````python
# notebooks can start in wa, scripts, or homework; let's find the course folder
course_folder = Path.cwd().resolve()
if course_folder.name in {'scripts', 'homework'}:
    course_folder = course_folder.parent
if not (course_folder / 'pyproject.toml').is_file():
    raise RuntimeError('Open the wa course folder in VS Code, then restart this notebook.')

# load our .env file; an existing environment variable can also supply the key
load_dotenv(course_folder / '.env')
my_api_key = os.getenv('OPENAI_API_KEY')

if not my_api_key or not my_api_key.strip():
    raise RuntimeError(
        'No OpenAI API key found. Add OPENAI_API_KEY to the .env file in your wa folder, '
        'save the file, then rerun this cell. Do not paste your key into this notebook.'
    )

# check that we found a key WITHOUT printing it
print('Successfully retrieved OpenAI API key!')
````

Let's create an OpenAI `client`. Think of it as our connection to the API: it knows which key to use and gives us methods for making requests.

We'll wait up to 30 seconds for a response and turn off automatic retries. If something goes wrong, we'll see the error rather than silently repeat the request.

````python
client = OpenAI(api_key=my_api_key, timeout=30, max_retries=0)
````

## 4.2 Make an API call

Now for the fun part! Let's ask `gpt-5.4-nano` for a short joke. It's a small, inexpensive model; we don't need a large model for this.

The function below sends our message and returns the answer as text. `max_completion_tokens=128` limits the length of the response, and `reasoning_effort='none'` avoids extra reasoning for this simple request.

Model names and prices can change. You can check the [model's current details here](https://developers.openai.com/api/docs/models/gpt-5.4-nano).

Running the last cell makes ONE paid API request. Running it again makes another one, so don't keep rerunning it by accident!

````python
def make_api_call(model, prompt):
    response = client.chat.completions.create(
        model=model,
        messages=[
            {'role': 'user', 'content': prompt}
        ],
        reasoning_effort='none',
        max_completion_tokens=128
    )
    return response.choices[0].message.content
````

````python
model_name = 'gpt-5.4-nano'
joke = make_api_call(model_name, 'Tell me one short joke, in at most two sentences.')
print(joke)
````
